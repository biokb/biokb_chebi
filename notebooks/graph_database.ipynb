# Graph database: import, query & configure

This notebook demonstrates how to import data into the Neo4j graph database, query the graph, and configure the database settings.

Before starting the import into neo4j start a test container with by copying and running the following command in terminal:
```bash
podman run --rm --name biokb-neo4j-test -p7474:7474 -p7687:7687 -e NEO4J_AUTH=neo4j/neo4j_password neo4j:5.26.31
```

What the last command does is start a Neo4j test container in the background, exposing the necessary ports for the Neo4j browser and Bolt protocol, and sets the authentication credentials to `neo4j/neo4j_password`.

````python
import os

os.environ.pop(
    "CONNECTION_STR", None
)  # to make sure the default CONNECTION_STR is used
````

### Create RDF turtle files

To import data into the Neo4j graph database, you first need to create an RDF turtle files containing nodes and edges in the appropriate format.

````python
from biokb_chebi import create_ttls

create_ttls()
````

### Import RDF turtle files into Neo4j

In the terminal, you can run the CLI commands to import data, create TTL files, and import them into the Neo4j graph database.

```bash
biokb_chebi import-neo4j
```


Or, you can then run the following cell to import the turtle files into the Neo4j graph database using the `import_neo4j` function.

````python
from biokb_chebi import import_neo4j

import_neo4j()
````

You can now access the Neo4j browser at [http://localhost:7474](http://localhost:7474) using the credentials `neo4j/neo4j_password`.

if needed you can also specify the uri, user and password for the Neo4j database when importing the turtle files. For example:
```python
from biokb_chebi import import_neo4j

import_neo4j(
    neo4j_uri="bolt://localhost:7687", 
    neo4j_user="neo4j", 
    neo4j_pwd="neo4j_password"
)
```

### using .env

You can create a (or add to if already exists) `.env` file the same directory :
```
NEO4J_USER=neo4j
NEO4J_PASSWORD=neo4j_password
NEO4J_HOST=localhost
NEO4J_URI=bolt://${NEO4J_HOST}:7687
```

Configuration will be automatically loaded from the `.env` file when you call `import_neo4j()`.

```python
from biokb_chebi import import_neo4j
import_neo4j()
```
